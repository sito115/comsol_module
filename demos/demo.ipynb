### Define data folders

```python
root = Path().resolve().parent
data_folder = root / "tests/data"
file = data_folder / "Example_Transient_NonSweep.vtu"
```

## Load vtu-file

```python
comsol_data = ComsolVtu.from_file(file)
comsol_data.info()
```

### Load temperature field as point data

```python
T = comsol_data.get_array("Temperature")
T.shape  # (time steps, points)
```

### Load temperature field as cell data

```python
comsol_data.convert_to_cell_data()
T = comsol_data.get_array("Temperature", location="cell")
T.shape
```

## Plot data above a certaint threshold

```python
field_name = comsol_data.format_field("Temperature", -1)
print(field_name)
mesh_cell_data = comsol_data.mesh.point_data_to_cell_data()
filtered_mesh = mesh_cell_data.threshold(
    scalars=field_name, value=400
)  # temperature above 480 K
filtered_mesh.plot(scalars=field_name, preference="cell")
```
